```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

data = {
    'Ogrenci': ['Ahmet', 'Mehmet', 'Ayse', 'Fatma', 'Ali', 'Veli'],
    'Bolum': ['BT', 'BT', 'YBS', 'YBS', 'BT', 'YBS'],
    'Puan': [60.0, np.nan, 80.0, 100.0, np.nan, 90.0]
}

df_temp = pd.DataFrame(data)
df_temp.to_csv('ogrenci_puanlari.csv', index=False)

df = pd.read_csv('ogrenci_puanlari.csv')
print("Ham Veri:")
display(df)

medyan_puan = df['Puan'].median()
df['Puan'] = df['Puan'].fillna(medyan_puan)
print(f"\nEksik puanlar medyan ({medyan_puan}) ile dolduruldu:")
display(df)

df['Gecti'] = df['Puan'].apply(lambda x: 1 if x >= 70 else 0)
print("\nGeçti (>=70) Sütunu Eklendi:")
display(df)

bolum_ortalamalari = df.groupby('Bolum')['Puan'].mean().reset_index()
print("\nBölüm Ortalamaları:")
display(bolum_ortalamalari)

plt.figure(figsize=(8, 5))
plt.bar(bolum_ortalamalari['Bolum'], bolum_ortalamalari['Puan'], color=['skyblue', 'salmon'])
plt.title('Bölüm Ortalama Puanları')
plt.xlabel('Bölüm')
plt.ylabel('Ortalama Puan')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```
